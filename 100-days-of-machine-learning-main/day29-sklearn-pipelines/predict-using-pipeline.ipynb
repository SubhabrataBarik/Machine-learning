# Day 29 — Predicting on New Data (With Pipeline)

## The Power of sklearn Pipelines
A **Pipeline** chains preprocessing + model into a single object.

At prediction time: just call `pipe.predict(raw_input)`.
The pipeline applies all preprocessing steps automatically in sequence.

**vs. manual approach:** Previously required loading 2 OHE objects, transforming separately, and concatenating manually.

## How It Works
```python
pipe = Pipeline([
    ('preprocessor', ColumnTransformer(...)),
    ('classifier', DecisionTreeClassifier())
])
pipe.fit(X_train, y_train)

# At prediction — one call, raw input:
pipe.predict(new_input)
```
