# Bagging vs Random Forest
Understanding the **key difference** between these two ensemble methods comes down to one thing:

| | Bagging | Random Forest |
|---|---|---|
| **Row sampling** | Yes — bootstrap (with replacement) | Yes — bootstrap (with replacement) |
| **Feature sampling** | Optional (`max_features`) — applied at the **tree level** | Always — applied at **every split node** |
| **Correlation between trees** | Higher (trees can use all features at each split) | Lower (each split is forced to consider only a random subset) |
| **Result** | Good variance reduction | Better variance reduction — trees are more diverse |

**Bottom line:** Random Forest is a special case of Bagging where feature randomness is injected at **every node**, not just at the tree level.


## Step 1: Imports

````python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier,RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
````

## Step 2: Generate Dataset
`make_classification` creates a synthetic binary classification dataset.
- `n_features=5` — 5 input features
- `n_informative=5` — all 5 features carry useful signal
- `n_redundant=0` — no redundant (linear combination) features
- `n_clusters_per_class=1` — one cluster per class (clean separation)


````python
X,y = make_classification(n_features=5, n_redundant=0, n_informative=5,n_clusters_per_class=1)
````

````python
df = pd.DataFrame(X,columns=['col1','col2','col3','col4','col5'])
df['target'] = y
print(df.shape)
df.head()
````

---
## Part 1: Bagging Classifier

### How Bagging Works
1. Draw a **bootstrap sample** (rows with replacement) from the training data
2. Train a full Decision Tree on that sample
3. At each split, the tree can consider **all features** (or `max_features` if set)
4. Repeat for `n_estimators` trees
5. Final prediction = **majority vote**

Setting `max_features=2` limits each tree to 2 features — but this is decided **once per tree**, not per split.


### Train the Bagging Classifier

````python
bag = BaggingClassifier(max_features=2)
````

````python
bag.fit(df.iloc[:,:5],df.iloc[:,-1])
````

### Visualise Tree 0 from Bagging
Each node in this tree can split on **any of the 2 randomly chosen features** — but that choice was fixed for the whole tree.
The tree tends to grow deeper and use the same dominant features repeatedly.


````python
plt.figure(figsize=(12,12))
plot_tree(bag.estimators_[0])
plt.show()
````

---
## Part 2: Random Forest Classifier

### How Random Forest Differs from Bagging
Random Forest adds **one extra layer of randomness** on top of Bagging:
- At **every single split node**, a fresh random subset of `max_features` features is considered
- This means even within one tree, different nodes use different feature subsets
- Result: trees are **less correlated** with each other → better ensemble diversity → lower variance

`max_features=2` here means: at each split, randomly pick 2 out of 5 features and find the best split among those 2.


### Train the Random Forest

````python
rf = RandomForestClassifier(max_features=2)
````

````python
rf.fit(df.iloc[:,:5],df.iloc[:,-1])
````

### Visualise Tree 4 from Random Forest
Notice how this tree looks different from the Bagging tree:
- Features used at each node vary more — the randomness is **per-split**, not per-tree
- The tree is typically shallower and less overfit on any single feature


````python
plt.figure(figsize=(12,12))
plot_tree(rf.estimators_[4])
plt.show()
````

---
## Summary: Key Difference Visualised

```
Bagging:
  Tree 1: [bootstrap sample] → grow tree, each split uses any of 2 fixed features
  Tree 2: [bootstrap sample] → grow tree, each split uses any of 2 fixed features
  ...

Random Forest:
  Tree 1: [bootstrap sample] → grow tree
    ├── Node 1 split: randomly pick 2 features → best split
    ├── Node 2 split: randomly pick 2 NEW features → best split
    └── Node 3 split: randomly pick 2 NEW features → best split
  Tree 2: same process with different bootstrap sample
  ...
```

**Random Forest = Bagging + per-node feature randomness**

This extra randomness de-correlates the trees, which is why Random Forest usually outperforms plain Bagging.
