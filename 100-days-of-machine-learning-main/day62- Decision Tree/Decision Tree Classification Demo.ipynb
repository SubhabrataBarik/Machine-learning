# Decision Tree Classifier — Demo
This notebook demonstrates a Decision Tree on two datasets:
1. **Iris** — classic multi-class classification (3 species)
2. **Social Network Ads** — binary purchase prediction (bought/not bought)

### How a Decision Tree works
- Repeatedly splits the data on the feature that most reduces **Gini Impurity**
- Each leaf node makes a final class prediction
- Easy to visualise and interpret — you can trace exactly why a prediction was made


---
## Part 1: Iris Dataset
### Step 1: Load Iris
4 features: sepal length, sepal width, petal length, petal width.
3 classes: Setosa (0), Versicolor (1), Virginica (2).


```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

### Step 2: Prepare Features and Target

```python
X=iris.data
```

```python
y=iris.target
y
```

### Step 3: Train/Test Split — 80/20

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(X,y, test_size=0.2)
```

```python
X_train.shape
```

```python
X_test.shape
```

### Step 4: Train Decision Tree
Default `DecisionTreeClassifier` — no depth limit, grows until all leaves are pure.
May overfit on noisy datasets but works well on clean ones like Iris.


```python
from sklearn.tree import DecisionTreeClassifier
```

```python
clf=DecisionTreeClassifier()
```

```python
clf.fit(X_train,y_train)
```

### Step 5: Predict and Evaluate

```python
y_pred=clf.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

### Step 6: Visualise the Tree
`plot_tree` draws the full decision tree.
Each node shows: feature used, threshold, gini, samples, class distribution.
Large figure size set to make it readable.


```python
from sklearn.tree import plot_tree
```

```python
from matplotlib.pylab import rcParams
rcParams['figure.figsize'] = 80,50

plot_tree(clf)
```

## Part 2: Social Network Ads Dataset
Binary classification — predict whether a user purchased a product.

Features:
- `Gender` (encoded: Male=0, Female=1)
- `Age`
- `EstimatedSalary`

Target: `Purchased` (0 = No, 1 = Yes)


```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('Social_Network_Ads.csv')
data.head()
```

### Step 7: Encode Gender
`map({'Male': 0, 'Female': 1})` converts the string column to numeric so sklearn can use it.


```python
data['Gender'] = data['Gender'].map({'Male': 0, 'Female': 1})
```

### Step 8: Prepare Features and Target

```python
X=data.iloc[:, 1:4].values
y=data.iloc[:,-1].values
```

```python
X.shape
```

```python
y.shape
```

### Step 9: Train with max_depth=3
`max_depth=3` limits the tree to 3 levels — prevents overfitting and keeps the visualisation readable.
Deeper trees memorise the training data; shallower trees generalise better.


```python
clf1=DecisionTreeClassifier(max_depth=3)
```

```python
clf1.fit(X,y)
```

### Step 10: Visualise the Depth-Limited Tree
Compared to the Iris tree, this one is much smaller and more interpretable.
Trace any path from root to leaf to understand a prediction.


```python
rcParams['figure.figsize'] = 80,50

plot_tree(clf1)
```
