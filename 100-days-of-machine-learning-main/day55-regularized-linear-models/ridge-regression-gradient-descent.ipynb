# Day 55 — Ridge Regression via Gradient Descent

## Ridge GD vs Closed Form
Closed form: `β = (XᵀX + αI)⁻¹ Xᵀy` — exact but O(p³) to invert.
Gradient Descent: iterative, scales to large p, same result.

## Ridge Gradient
The Ridge loss:
```
L = MSE + α · Σβᵢ²
```

Gradient for each coefficient:
```
∂L/∂βⱼ = ∂MSE/∂βⱼ + 2α·βⱼ
```

This shrinks each coefficient by `2α·βⱼ` at each step — hence the name **weight decay** in deep learning.

## Approaches Compared
1. sklearn `SGDRegressor(penalty='l2')` — built-in
2. sklearn `Ridge(solver='sparse_cg')` — iterative solver
3. `MeraRidgeGD` — our custom implementation


## Step 1: Imports + Load Data

````python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
import numpy as np
````

````python
X,y = load_diabetes(return_X_y=True)
````

````python
from sklearn.model_selection import train_test_split
````

````python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=4)
````

## Step 2: SGDRegressor with L2 Penalty

````python
from sklearn.linear_model import SGDRegressor
````

````python
reg = SGDRegressor(penalty='l2',max_iter=500,eta0=0.1,learning_rate='constant',alpha=0.001)
````

````python
reg.fit(X_train,y_train)

y_pred = reg.predict(X_test)
print("R2 score",r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
````

## Step 3: Ridge with Sparse CG Solver (iterative)

````python
from sklearn.linear_model import Ridge

reg = Ridge(alpha=0.001, max_iter=500,solver='sparse_cg')
````

````python
reg.fit(X_train,y_train)

y_pred = reg.predict(X_test)
print("R2 score",r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
````

## Step 4: MeraRidgeGD — Custom Ridge via Gradient Descent

````python
class MeraRidgeGD:
    
    def __init__(self,epochs,learning_rate,alpha):
        
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.alpha = alpha
        self.coef_ = None
        self.intercept_ = None
        
    def fit(self,X_train,y_train):
        
        self.coef_ = np.ones(X_train.shape[1])
        self.intercept_ = 0
        thetha = np.insert(self.coef_,0,self.intercept_)
        
        X_train = np.insert(X_train,0,1,axis=1)
        
        for i in range(self.epochs):
            thetha_der = np.dot(X_train.T,X_train).dot(thetha) - np.dot(X_train.T,y_train) + self.alpha*thetha
            thetha = thetha - self.learning_rate*thetha_der
        
        self.coef_ = thetha[1:]
        self.intercept_ = thetha[0]
    
    def predict(self,X_test):
        
        return np.dot(X_test,self.coef_) + self.intercept_
````

## Step 5: Train and Evaluate

````python
reg = MeraRidgeGD(epochs=500,alpha=0.001,learning_rate=0.005)
````

````python
reg.fit(X_train,y_train)

y_pred = reg.predict(X_test)
print("R2 score",r2_score(y_test,y_pred))
print(reg.coef_)
print(reg.intercept_)
````
