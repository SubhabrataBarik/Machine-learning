# Day 55 — Ridge Regression (L2 Regularization)

## Why Regularize?
Linear regression minimises MSE — but with many features or correlated features, coefficients can become very large (overfitting).

**Regularization** adds a penalty to the loss function to keep coefficients small:

```
Ridge Loss = MSE + α · Σβᵢ²
           = Σ(y - ŷ)² + α · Σβᵢ²
```

## L2 Penalty (Ridge)
- Penalises the **squared magnitude** of coefficients
- Large `α` → stronger penalty → coefficients shrink toward (but never exactly) zero
- Small `α = 0` → ordinary linear regression

## Effect of α
| α | Effect |
|---|--------|
| 0 | Standard OLS (no regularization) |
| Small | Slight regularization |
| Large | Strong shrinkage — simpler model |
| Very large | All coefficients → 0 (predict mean) |

## When to Use Ridge
- Many correlated features (multicollinearity)
- When all features are somewhat useful (keep all, just shrink)
- vs Lasso: Ridge keeps all features; Lasso can zero some out

## Dataset: Diabetes (10 features)


## Step 1: Imports

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
````

## Step 2: Load Diabetes Dataset

````python
from sklearn.datasets import load_diabetes

data=load_diabetes()
````

````python
print(data.DESCR)
````

````python
X=data.data
y=data.target
````

## Step 3: Train/Test Split + Baseline Linear Regression

````python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=45)
````

````python
from sklearn.linear_model import LinearRegression
L=LinearRegression()
````

````python
L.fit(X_train,y_train)
````

````python
print(L.coef_)
print(L.intercept_)
````

````python
y_pred=L.predict(X_test)
````

````python
from sklearn.metrics import r2_score,mean_squared_error

print("R2 score",r2_score(y_test,y_pred))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred)))
````

## Step 4: Apply Ridge Regression (alpha=100000 — extreme)

````python
from sklearn.linear_model import Ridge
R=Ridge(alpha=100000)
````

````python
R.fit(X_train,y_train)
````

````python
print(R.coef_)
print(R.intercept_)
````

````python
y_pred1=R.predict(X_test)
````

````python
print("R2 score",r2_score(y_test,y_pred1))
print("RMSE",np.sqrt(mean_squared_error(y_test,y_pred1)))
````

## Step 5: Visualise Effect of α on Polynomial Degree-16 Model

````python
m = 100
x1 = 5 * np.random.rand(m, 1) - 2
x2 = 0.7 * x1 ** 2 - 2 * x1 + 3 + np.random.randn(m, 1)

plt.scatter(x1, x2)
plt.show()
````

### Compare Fits at Different α Values
Higher α → smoother line (less overfitting).
Very high α → underfitting (line barely moves).


````python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

def get_preds_ridge(x1, x2, alpha):
    model = Pipeline([
        ('poly_feats', PolynomialFeatures(degree=16)),
        ('ridge', Ridge(alpha=alpha))
    ])
    model.fit(x1, x2)
    return model.predict(x1)

alphas = [0, 20, 200]
cs = ['r', 'g', 'b']

plt.figure(figsize=(10, 6))
plt.plot(x1, x2, 'b+', label='Datapoints')

for alpha, c in zip(alphas, cs):
    preds = get_preds_ridge(x1, x2, alpha)
    # Plot
    plt.plot(sorted(x1[:, 0]), preds[np.argsort(x1[:, 0])], c, label='Alpha: {}'.format(alpha))

plt.legend()
plt.show()
````
