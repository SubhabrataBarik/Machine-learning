# AdaBoost — Manual Step-by-Step Implementation

AdaBoost (**Ada**ptive **Boost**ing) trains weak classifiers sequentially.
Each iteration focuses more on the samples the previous classifier got wrong.

### The Algorithm (3 iterations shown here)
```
1. Assign equal weights to all samples: w_i = 1/N
2. For each round t:
   a. Sample a new dataset using current weights (weighted bootstrap)
   b. Train a weak learner (decision stump, max_depth=1) on it
   c. Calculate model weight α_t = 0.5 × log((1 - error) / error)
      → Low error → high α (model gets more say)
      → High error → low α (model is nearly ignored)
   d. Update sample weights:
      → Correctly classified: weight × exp(-α)  (down-weighted)
      → Misclassified: weight × exp(+α)          (up-weighted)
   e. Normalise weights so they sum to 1
3. Final prediction = sign(Σ α_t × prediction_t)
   → Weighted majority vote of all weak learners
```


## Step 1: Create Dataset
10 points with 2 features (X1, X2) and binary labels (0 or 1).

````python
import pandas as pd
import numpy as np
from mlxtend.plotting import plot_decision_regions
````

````python
df = pd.DataFrame()
````

````python
df['X1'] = [1,2,3,4,5,6,6,7,9,9]
df['X2'] = [5,3,6,8,1,9,5,8,9,2]
df['label'] = [1,1,0,1,0,1,0,1,0,0]
````

````python
df
````

````python
import seaborn as sns
sns.scatterplot(x=df['X1'],y=df['X2'],hue=df['label'])
````

## Step 2: Initialise Equal Weights
Every sample starts with weight = 1/N = 1/10 = 0.1.
Higher weight = model pays more attention to this sample next round.


````python
df['weights'] = 1/df.shape[0]
````

````python
df
````

````python
from sklearn.tree import DecisionTreeClassifier
````

---
## Round 1 — Train First Stump
A **decision stump** = Decision Tree with `max_depth=1` (one split, two leaves).
It's a weak learner — just slightly better than random guessing.


````python
dt1 = DecisionTreeClassifier(max_depth=1)
````

````python
X = df.iloc[:,0:2].values
y = df.iloc[:,2].values
````

````python
# Step 2 - Train 1st model
dt1.fit(X,y)
````

````python
from sklearn.tree import plot_tree
plot_tree(dt1)
````

````python
plot_decision_regions(X, y, clf=dt1, legend=2)
````

### Round 1 — Predictions

````python
df['y_pred'] = dt1.predict(X)
````

````python
df
````

### Round 1 — Calculate Model Weight α₁
`α = 0.5 × log((1 - error) / error)`
- `error` = sum of weights of misclassified samples
- 3 misclassified out of 10 → error = 0.3
- α₁ = 0.5 × log(0.7/0.3) ≈ **0.423**
- This stump gets moderate influence in the final vote.


````python
def calculate_model_weight(error):

  return 0.5*np.log((1-error)/(error))
````

````python
# Step 3 - calculate model weight
alpha1 = calculate_model_weight(0.3)
alpha1
````

### Round 1 — Update Sample Weights
Misclassified samples get **higher** weight (model will focus on them next round).
Correctly classified samples get **lower** weight.

Formula:
- Correct: `w × exp(-α)`
- Wrong:   `w × exp(+α)`


````python
# Step 4 - Update weights
def update_row_weights(row,alpha=0.423):
  if row['label'] == row['y_pred']:
    return row['weights'] * np.exp(-alpha)
  else:
    return row['weights'] * np.exp(alpha)
````

````python
df['updated_weights'] = df.apply(update_row_weights,axis=1)
````

````python
df
````

````python
df['updated_weights'].sum()
````

### Round 1 — Normalise Weights
Divide each weight by the total so they sum to 1 again.

````python
df['nomalized_weights'] = df['updated_weights']/df['updated_weights'].sum()
````

````python
df
````

````python
df['nomalized_weights'].sum()
````

### Round 1 — Build Cumulative Weight Ranges
To do **weighted sampling**, assign each sample a range [lower, upper] proportional to its weight.
A random number in [0,1] falls into one sample's range → that sample is selected.
Higher-weight samples have wider ranges → more likely to be selected.


````python
df['cumsum_upper'] = np.cumsum(df['nomalized_weights'])
````

````python
df['cumsum_lower'] = df['cumsum_upper'] - df['nomalized_weights']
````

````python
df[['X1','X2','label','weights','y_pred','updated_weights','cumsum_lower','cumsum_upper']]
````

### Round 1 — Create New Weighted Dataset

````python
def create_new_dataset(df):

  indices = []

  for i in range(df.shape[0]):
    a = np.random.random()
    for index,row in df.iterrows():
      if row['cumsum_upper'] > a and a > row['cumsum_lower']:
        indices.append(index)
  return indices
````

````python
index_values = create_new_dataset(df)

index_values
````

````python
second_df = df.iloc[index_values,[0,1,2,3]]
````

````python
second_df
````

---
## Round 2 — Train Second Stump on Weighted Dataset
The new dataset over-represents the samples Round 1 got wrong.
The second stump will focus on getting those right.


````python
dt2 = DecisionTreeClassifier(max_depth=1)
````

````python
X = second_df.iloc[:,0:2].values
y = second_df.iloc[:,2].values
````

````python
dt2.fit(X,y)
````

````python
plot_tree(dt2)
````

````python
plot_decision_regions(X, y, clf=dt2, legend=2)
````

### Round 2 — Predictions and Model Weight α₂
Only 1 misclassified → error = 0.1 → α₂ ≈ 1.09 (much higher influence than Round 1).

````python
second_df['y_pred'] = dt2.predict(X)
````

````python
second_df
````

````python
alpha2 = calculate_model_weight(0.1)
````

````python
alpha2
````

### Round 2 — Update and Normalise Weights

````python
# Step 4 - Update weights
def update_row_weights(row,alpha=1.09):
  if row['label'] == row['y_pred']:
    return row['weights'] * np.exp(-alpha)
  else:
    return row['weights'] * np.exp(alpha)
````

````python
second_df['updated_weights'] = second_df.apply(update_row_weights,axis=1)
````

````python
second_df
````

````python
second_df['nomalized_weights'] = second_df['updated_weights']/second_df['updated_weights'].sum()
````

````python
second_df
````

````python
second_df['nomalized_weights'].sum()
````

````python
second_df['cumsum_upper'] = np.cumsum(second_df['nomalized_weights'])
````

````python
second_df['cumsum_lower'] = second_df['cumsum_upper'] - second_df['nomalized_weights']
````

### Round 2 — Create Third Weighted Dataset

````python
index_values = create_new_dataset(second_df)
````

````python
third_df = second_df.iloc[index_values,[0,1,2,3]]
````

````python
third_df
````

## Round 3 — Train Third Stump


````python
dt3 = DecisionTreeClassifier(max_depth=1)

X = second_df.iloc[:,0:2].values
y = second_df.iloc[:,2].values

dt3.fit(X,y)
````

````python
plot_decision_regions(X, y, clf=dt3, legend=2)
````

### Round 3 — Model Weight α₃
7 misclassified → error = 0.7 → α₃ ≈ **-0.42** (negative! this stump is inverted — worse than random).

````python
alpha3 = calculate_model_weight(0.7)
alpha3
````

````python
print(alpha1,alpha2,alpha3)
````

---
## Final Prediction — Weighted Majority Vote
```
Final = sign(α₁ × pred₁ + α₂ × pred₂ + α₃ × pred₃)
```
- Positive result → class 1
- Negative result → class 0 (since labels are mapped to +1/-1 internally)

Note: Labels 0 → -1 for the sign computation.
