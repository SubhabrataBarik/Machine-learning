# Day 57 — ElasticNet Regression (L1 + L2 Combined)

## What is ElasticNet?
ElasticNet combines **both** L1 (Lasso) and L2 (Ridge) penalties:

```
ElasticNet Loss = MSE + α · l1_ratio · Σ|βᵢ| + α · (1-l1_ratio)/2 · Σβᵢ²
```

## l1_ratio Parameter
| l1_ratio | Behaviour |
|----------|-----------|
| 0 | Pure Ridge |
| 1 | Pure Lasso |
| 0.5 | Balanced mix |
| 0.9 | Mostly Lasso with L2 stability |

## Why ElasticNet?
Lasso has issues when:
- Features are **highly correlated** — Lasso arbitrarily picks one, Ridge keeps both
- p >> n — Lasso selects at most n features

ElasticNet combines:
- L1's sparsity (feature selection)
- L2's grouping effect (keeps correlated features together)

## Best of Both Worlds
In practice, `l1_ratio=0.9` often works well: mostly Lasso's sparsity but with Ridge's stability.

## Dataset: Diabetes (10 features)


## Step 1: Imports

````python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
````

## Step 2: Load Diabetes Dataset + Split

````python
X,y = load_diabetes(return_X_y=True)
````

````python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
````

## Step 3: Baseline — Standard Linear Regression

````python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
````

## Step 4: Ridge Regression (alpha=0.1)

````python
# Ridge 
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
````

## Step 5: Lasso Regression (alpha=0.01)

````python
# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
````

## Step 6: ElasticNet (alpha=0.005, l1_ratio=0.9)

````python
# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
````
